## Example 1 - Single fit
Single fitter, in a new canvas.

## Example 2 - Multi Histogram Fit
Two separate fitters, on different histograms, run in different subplot axes.

````python
%matplotlib notebook
fix,ax = plt.subplots(1,2,figsize=(10,4))
fp = []
countsList = [counts0, counts1]
for i in range(2):
    fp.append(
        BiGausFitter(
            countsList[i], edges,
            dofit=True, dohesse=True,
            title = 'Hist %i'%i,
            axis_ = ax[i]
        )
    )
````

## Example Summary Plots
1. FOM vs. Index and
2. Right Peak Counts vs. Index

````python
%matplotlib inline

fom_all = [ f.FOM()  for f in fp ]
right_counts_all = [  f.peak_counts()[1]  for f in fp  ]

fig,ax = plt.subplots(1,2,figsize=(10,4))
ax[0].errorbar(
    arange(0,len(fp),1,int),
    [x.nominal_value for x in fom_all],
    [x.std_dev for x in fom_all]
    ,ls = 'none', marker='s', label = 'FOM vs. Index'
)
ax[1].errorbar(
    arange(0,len(fp),1,int),
    [x.nominal_value for x in right_counts_all],
    [x.std_dev for x in right_counts_all]
    ,ls = 'none', marker='s', label = 'Counts in Right Peak vs. Index'
)
plt.show()
````
